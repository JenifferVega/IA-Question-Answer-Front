````python
from txtai.embeddings import Embeddings
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import numpy as np

# Initialize txtai embeddings
embeddings = Embeddings()

# Generate embeddings for the texts
vectors = [embeddings.transform(text) for text in texts]

print(vectors)

# Convert list of vectors to numpy array
vectors = np.array(vectors).squeeze()

# Reduce dimensionality with PCA
pca = PCA(n_components=2)
reduced_vectors = pca.fit_transform(vectors)

# Use K-means to cluster the reduced embeddings
kmeans = KMeans(n_clusters=2, random_state=42).fit(reduced_vectors)
classes = kmeans.labels_

classes
````

````python
# Validate clustering with silhouette score
score = silhouette_score(reduced_vectors, classes)
print(f"Silhouette Score: {score}")

# Create a dictionary to hold clusters
clusters = {}

# Populate the dictionary with texts grouped by labels
for text, label in zip(texts, classes):
    if label not in clusters:
        clusters[label] = []
    clusters[label].append(text)

# Access clusters by their label
for label, cluster_texts in clusters.items():
    print(f"Cluster {label}: {cluster_texts}")
````

````python
#this is possible to do it using embeddings pipeline

from sentence_transformers import SentenceTransformer, util

# Load a pre-trained sentence transformer model (e.g., 'all-MiniLM-L6-v2')
model = SentenceTransformer('nli-mpnet-base-v2')

# Example sentences
sentence_a = "Contact details and department or work area of operation."
sentence_b = "Template 2: Advantages and necessity of organisational performance development."

# Step 1: Generate embeddings for each sentence
embedding_a = model.encode(sentence_a, convert_to_tensor=True)
embedding_b = model.encode(sentence_b, convert_to_tensor=True)

# Step 2: Compute cosine similarity between the two embeddings
similarity_score = util.pytorch_cos_sim(embedding_a, embedding_b)

# Step 3: Print similarity score
print(f"Similarity score: {similarity_score.item()}")

# Step 4: Check if similarity is above a certain threshold (e.g., 0.5 for related topics)
if similarity_score > 0.5:
    print("The sentences are likely part of the same topic.")
else:
    print("The sentences are likely part of different topics.")
````

````python
total_text_segments = text_segments[0]
transition_indices = []

# Set a similarity threshold for detecting topic changes
SIMILARITY_THRESHOLD = 0.5

# Initialize the first section recognized
sections_recognized = [total_text_segments[0]]

# Loop through the segments and compare each with the previous one
for i in range(1, len(total_text_segments)):
    # Compare the current segment with the previous one
    current_segment = total_text_segments[i]
    previous_segment = total_text_segments[i - 1]

    # Calculate the similarity between the current and previous segments
    similarity_score = embeddings.similarity(previous_segment, [current_segment])[0][1]

    # Check if the similarity score is below the threshold
    if similarity_score < SIMILARITY_THRESHOLD:
        print(f"*** Topic change detected between segment {i-1} and segment {i} ***")
        #print(current_segment[0:100])
        transition_indices.append(i)
````

````python
transition_indices
````

````python
# Now extract sections based on the transition indices
sections = []
start_index = 0

# Iterate over transition indices to create sections
for transition_index in transition_indices:
    # Extract the section
    section = total_text_segments[start_index:transition_index]
    # Combine the segments into a single string (optional, for readability)
    combined_section = " ".join(section)
    sections.append(combined_section)
    # Update the start index for the next section
    start_index = transition_index

# Add the last section after the last transition
last_section = total_text_segments[start_index:]
combined_last_section = " ".join(last_section)
sections.append(combined_last_section)

# Print the extracted sections
for idx, section in enumerate(sections):
    print(f"Section {idx + 1}:\n{section}\n")
````

````python
# Function to cluster texts and extract sections
def cluster_texts_in_batches(texts, batch_size=10, n_clusters=2):
    sections = []  # List to store sections
    i = 0  # Starting index
    
    while i < len(texts):
        # Get the current batch of texts (from i to i + batch_size)
        batch_texts = texts[i:i + batch_size]

        # Handle the case where batch_texts contains only one text
        if len(batch_texts) == 1:
            sections.append(batch_texts)
            i += 1
            continue

        # Generate embeddings for the batch
        vectors = np.array([embeddings.transform(text) for text in batch_texts]).squeeze()

        # Reduce dimensionality with PCA
        pca = PCA(n_components=2)
        reduced_vectors = pca.fit_transform(vectors)

        # Cluster the reduced embeddings with KMeans
        kmeans = KMeans(n_clusters=n_clusters, random_state=42).fit(reduced_vectors)
        labels = kmeans.labels_

        # Find the first transition point (where the cluster changes)
        transition_index = None
        for j in range(1, len(labels)):
            if labels[j] != labels[j - 1]:
                transition_index = j
                break

        # If no transition point, treat the entire batch as one section
        if transition_index is None:
            sections.append(texts[i:i + batch_size])
            i += batch_size
        else:
            # Include the entire same-cluster group into one section
            sections.append(texts[i:i + transition_index])
            # Move the starting point to the transition point
            i += transition_index
    
    return sections
````

````python
# Process the text segments
sections = cluster_texts_in_batches(total_text_segments)
````

````python
import re

def clean_text(text):
    # Step 1: Normalize newlines and spaces
    text = re.sub(r'\n+', '\n', text)  # Replace multiple newlines with a single newline
    text = re.sub(r' +', ' ', text)  # Replace multiple spaces with a single space
    
    # Step 2: Remove special formatting and table structures
    text = re.sub(r'\|---\|', '', text)
    text = re.sub(r'\|', '', text)
    
    # Step 3: Strip headers and repetitive phrases
    #text = re.sub(r'Template \d+:.*\n', '', text)  # Remove template headers
    #text = re.sub(r'Fill this in.*\n', '', text)  # Remove filler instructions
    
    # Step 4: Segment based on section headers
    text = re.sub(r'(Template \d+:|Checklist \d+:)', r'\n=== \1 ===\n', text)  # Add markers for segmentation
    
    return text

# Example usage
cleaned_text = clean_text(sections_extracted[0])
print(cleaned_text)
````

````python
def run_prompt(llm, question, text):
    prompt = f"""system
    You are a friendly assistant. You answer questions from users.
    user
    Answer the following question using only the context below. Only include information specifically discussed.

    question: {question}
    context: {text} 
    assistant
    """
    
    if isinstance(llm, LLM):
        # Handle LLMs initialized from txtai pipeline
        return llm(prompt, maxlength=4096)
    else:
        # Handle custom ollama LLM
        return llm(prompt)
````

````python
# GPT-4 LLM class definition
class GPT4LLM:
    def __init__(self, client):
        self.client = client
        
    def __call__(self, prompt):
        response = self.client.chat.completions.create(
            model="gpt-4o-mini",  # Specify the GPT-4 model
            messages=[
                {"role": "system", "content": "You are a helpful assistant."},
                {"role": "user", "content": prompt}
            ],
            max_tokens=2048,  # Adjust as necessary
            temperature=0.7    # Adjust the creativity/randomness level
        )
        response_dict = response.to_dict()  # Convert the custom response object to a dictionary

        # Access the response content
        return response_dict['choices'][0]['message']['content'].strip()
        

# Function to fetch context based on a question using txtai embeddings
def context(question):
    search_results = embeddings.search(question)
    context = "\n".join(x["text"] for x in search_results)
    return context

# Function to create a prompt and call the GPT-4 LLM
def run_prompt_gpt4(llm, question, text):
    prompt = f"""Answer the following question using only the context below. Only include information specifically discussed.
    
    question: {question}
    context: {text} 
    """
    return llm(prompt)

# Function to perform retrieval-augmented generation (RAG)
def rag(llm, question):
    text_context = context(question)  # Get the context from the embedding search
    return run_prompt_gpt4(llm, question, text_context)  # Send the prompt with context to the LLM
````

````python
import json

def parse_gpt_output_to_json(gpt_output):
   
    print("gpt_output",gpt_output)
   
    cleaned_output = gpt_output.strip()
    start_idx = cleaned_output.find('```json')
    end_idx = cleaned_output.rfind('```')

    if start_idx == -1 or end_idx == -1 or start_idx >= end_idx:
        print("No valid JSON markers found in the text.")
        return None

    json_string = cleaned_output[start_idx + len('```json'):end_idx].strip()

    match = re.search(r'\[\s*{.*?}\s*\]', json_string, re.DOTALL)
    if match:
        json_str = match.group(0)
        try:
            product_info = json.loads(json_str)
            return product_info
        except json.JSONDecodeError as e:
            print(f"Failed to decode JSON: {e}")
            return None
    else:
        print("No valid JSON found in the text.")
        return None
````

````python
def split_paragraph_in_two(paragraph, search_text):
    # Find the position of the search_text in the paragraph
    position = paragraph.find(search_text)
    
    # If the search_text is found, split the paragraph into two parts
    if position != -1:
        # The first part includes the search_text
        first_part = paragraph[:position + len(search_text)].strip()
        
        # The second part is everything after the search_text
        second_part = paragraph[position + len(search_text):].strip()
        
        return first_part, second_part
    else:
        return None, paragraph
````
